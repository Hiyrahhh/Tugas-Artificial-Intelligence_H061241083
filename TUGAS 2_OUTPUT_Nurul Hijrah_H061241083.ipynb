### **2. Estimasi Orientasi Sumbu Tegasan Utama**

Inversi tegasan (*stress inversion*) bertujuan untuk memperkirakan orientasi tiga sumbu tegasan utama, yaitu tegasan maksimum ($\sigma_1$), tegasan menengah ($\sigma_2$), dan tegasan minimum ($\sigma_3$).

Orientasi setiap sumbu tegasan dinyatakan menggunakan azimut (*azimuth*) dan sudut kemiringan garis (*plunge*). Ketiga sumbu tegasan utama saling tegak lurus satu sama lain.

Secara umum, hubungan ketiga sumbu tegasan dapat dinyatakan sebagai berikut:

$$
\boldsymbol{\sigma}_1 \perp \boldsymbol{\sigma}_2
$$

$$
\boldsymbol{\sigma}_2 \perp \boldsymbol{\sigma}_3
$$

$$
\boldsymbol{\sigma}_1 \perp \boldsymbol{\sigma}_3
$$

**Tabel Orientasi Sumbu Tegasan Utama**

| **Sumbu Tegasan** | **Azimut (°)** | **Plunge (°)** | **Keterangan** |
|:---:|---:|---:|---|
| $\sigma_1$ | Belum dihitung | Belum dihitung | Tegasan utama maksimum |
| $\sigma_2$ | Belum dihitung | Belum dihitung | Tegasan utama menengah |
| $\sigma_3$ | Belum dihitung | Belum dihitung | Tegasan utama minimum |

**Interpretasi:**

Orientasi sumbu tegasan utama digunakan untuk memahami arah gaya yang bekerja pada sistem sesar. Sumbu $\sigma_1$ menunjukkan arah tegasan utama maksimum, sedangkan sumbu $\sigma_3$ menunjukkan arah tegasan utama minimum. Nilai azimut dan *plunge* harus diperoleh melalui perhitungan inversi dari data mekanisme fokus gempa agar hasilnya dapat dipertanggungjawabkan secara ilmiah.


### **3. Interpretasi Rezim Tegasan**

Rezim tegasan menggambarkan kondisi tegasan yang mengontrol deformasi dan pergerakan sesar. Interpretasinya dilakukan dengan memperhatikan orientasi sumbu tegasan utama terhadap bidang horizontal dan vertikal.

**Klasifikasi Rezim Tegasan**

| **Rezim Tegasan** | **Orientasi Umum Sumbu Tegasan** | **Karakteristik** |
|---|---|---|
| Normal (*normal faulting*) | $\sigma_1$ cenderung vertikal dan $\sigma_3$ cenderung horizontal | Berkaitan dengan ekstensi atau peregangan kerak. |
| Sesar mendatar (*strike-slip faulting*) | $\sigma_2$ cenderung vertikal, sedangkan $\sigma_1$ dan $\sigma_3$ cenderung horizontal | Berkaitan dengan dominasi deformasi mendatar. |
| Sesar naik (*reverse/thrust faulting*) | $\sigma_1$ cenderung horizontal dan $\sigma_3$ cenderung vertikal | Berkaitan dengan kompresi atau pemendekan kerak. |

**Interpretasi:**

Data mekanisme fokus gempa yang memiliki komponen pergerakan mendatar dapat mengindikasikan aktivitas sesar mendatar (*strike-slip*). Namun, jenis pergerakan pada masing-masing event tidak cukup untuk memastikan rezim tegasan regional secara keseluruhan.

Penentuan rezim tegasan pada wilayah penelitian perlu didasarkan pada hasil inversi tegasan dari seluruh data mekanisme fokus gempa. Orientasi sumbu $\sigma_1$, $\sigma_2$, dan $\sigma_3$ selanjutnya digunakan untuk mengevaluasi apakah medan tegasan didominasi oleh kompresi, ekstensi, atau pergeseran mendatar.

### **4. Evaluasi Kualitas Hasil Inversi**

Kualitas hasil inversi tegasan dievaluasi dengan membandingkan arah slip hasil pengamatan dengan arah slip yang diprediksi oleh model tegasan. Salah satu parameter evaluasi yang digunakan adalah *misfit* sudut, yaitu sudut antara kedua vektor slip tersebut.

*Misfit* sudut dapat dihitung menggunakan persamaan berikut:

$$
\theta_i = \cos^{-1}\left(
\frac{
\mathbf{s}_{i,\mathrm{obs}} \cdot \mathbf{s}_{i,\mathrm{pred}}
}{
\left\|\mathbf{s}_{i,\mathrm{obs}}\right\|
\left\|\mathbf{s}_{i,\mathrm{pred}}\right\|
}
\right)
$$

**Keterangan:**

- $\theta_i$ = *misfit* sudut untuk event ke-$i$.
- $\mathbf{s}_{i,\mathrm{obs}}$ = vektor slip hasil pengamatan.
- $\mathbf{s}_{i,\mathrm{pred}}$ = vektor slip hasil prediksi model tegasan.
- $\cdot$ = operasi perkalian titik (*dot product*).
- $\cos^{-1}$ = fungsi invers kosinus.

Nilai *misfit* yang kecil menunjukkan bahwa arah slip hasil pengamatan mendekati arah slip yang diprediksi oleh model tegasan. Perhitungan perlu memperhatikan konvensi arah slip dan ambiguitas bidang nodal pada mekanisme fokus gempa.

### **5. Kesimpulan**

Berdasarkan parameter mekanisme fokus gempa yang tersedia, delapan event menunjukkan indikasi dominasi pergerakan sesar mendatar (*strike-slip*). Data tersebut dapat digunakan sebagai dasar untuk memperkirakan orientasi medan tegasan aktif melalui metode *stress inversion*.

Meskipun demikian, orientasi numerik sumbu tegasan \(\sigma_1\), \(\sigma_2\), dan \(\sigma_3\), penentuan rezim tegasan, serta evaluasi *misfit* harus diperoleh melalui perhitungan yang konsisten menggunakan seluruh data mekanisme fokus gempa.

Hasil inversi selanjutnya dapat digunakan untuk menginterpretasikan hubungan antara aktivitas seismik dan kondisi medan tegasan pada sistem sesar yang diteliti.

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import differential_evolution

# =====================================================
# 1. INPUT DATA
# =====================================================
df = pd.DataFrame({
    "Event": ["E1", "E2", "E3", "E4",
              "E5", "E6", "E7", "E8"],
    "Strike": [25, 31, 148, 37, 152, 28, 145, 34],
    "Dip":    [65, 61, 68, 63, 70, 67, 65, 60],
    "Rake":   [170, 175, 8, 168, 12, 172, 5, 178]
})

# =====================================================
# 2. KLASIFIKASI RAKE
# Konvensi: 0° = geser kanan,
# +90° = naik/reverse, ±180° = geser kiri,
# -90° = turun/normal.
# =====================================================
def classify_rake(rake):
    r = ((rake + 180) % 360) - 180

    if abs(r) >= 150:
        return "Geser kiri dominan"
    elif abs(r) <= 30:
        return "Geser kanan dominan"
    elif 30 < r < 150:
        return "Oblique reverse"
    else:
        return "Oblique normal"

df["Jenis Sesar"] = df["Rake"].apply(classify_rake)
print("DATA DAN KLASIFIKASI SESAR")
print(df.to_string(index=False))

# =====================================================
# 3. VEKTOR NORMAL BIDANG DAN ARAH SLIP
# Koordinat: East, North, Up
# =====================================================
def plane_vectors(strike, dip, rake):
    s, d, r = np.radians([strike, dip, rake])

    strike_vec = np.array([
        np.sin(s), np.cos(s), 0.0
    ])

    dip_vec = np.array([
        np.cos(d) * np.cos(s),
        -np.cos(d) * np.sin(s),
        -np.sin(d)
    ])

    normal = np.cross(strike_vec, dip_vec)
    normal /= np.linalg.norm(normal)

    # Rake positif diarahkan ke atas bidang
    slip = np.cos(r) * strike_vec - np.sin(r) * dip_vec
    slip /= np.linalg.norm(slip)

    return normal, slip

vectors = [
    plane_vectors(row.Strike, row.Dip, row.Rake)
    for row in df.itertuples()
]

normals = np.array([v[0] for v in vectors])
slips = np.array([v[1] for v in vectors])

# =====================================================
# 4. TENSOR TEGASAN DAN FUNGSI MISFIT
# Tensor simetris, trace = 0, dinormalisasi.
# Misfit = sudut antara slip observasi dan
# arah traksi geser yang diprediksi.
# =====================================================
def build_tensor(p):
    a, b, c, d, e = p

    T = np.array([
        [a, c, d],
        [c, b, e],
        [d, e, -a-b]
    ], dtype=float)

    norm = np.linalg.norm(T)
    if norm < 1e-12:
        return None

    return T / norm

def calculate_misfit(p):
    T = build_tensor(p)

    if T is None:
        return np.full(len(normals), np.pi)

    angles = []

    for n, observed_slip in zip(normals, slips):
        traction = T @ n
        shear = traction - np.dot(n, traction) * n

        magnitude = np.linalg.norm(shear)

        if magnitude < 1e-10:
            angles.append(np.pi)
            continue

        predicted_slip = shear / magnitude
        dot = np.clip(
            np.dot(predicted_slip, observed_slip),
            -1.0, 1.0
        )

        angles.append(np.arccos(dot))

    return np.array(angles)

def objective(p):
    angles = calculate_misfit(p)
    return np.mean(angles ** 2)

# Optimasi deterministik dengan seed tetap
result = differential_evolution(
    objective,
    bounds=[(-1, 1)] * 5,
    seed=42,
    tol=1e-10,
    popsize=20,
    maxiter=1500,
    polish=True
)

T = build_tensor(result.x)
misfit_deg = np.degrees(calculate_misfit(result.x))

df["Misfit (derajat)"] = misfit_deg

# =====================================================
# 5. ORIENTASI SUMBU UTAMA
# Eigenvalue diurutkan dari terbesar ke terkecil.
# =====================================================
eigenvalues, eigenvectors = np.linalg.eigh(T)
order = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[order]
eigenvectors = eigenvectors[:, order]

def azimuth_plunge(v):
    v = v.copy()

    # Pilih representasi sumbu pada hemisfer bawah
    if v[2] > 0:
        v = -v

    azimuth = np.degrees(
        np.arctan2(v[0], v[1])
    ) % 360

    plunge = np.degrees(
        np.arcsin(np.clip(-v[2], -1, 1))
    )

    return azimuth, plunge

axes = []

for i, name in enumerate(["Sigma1", "Sigma2", "Sigma3"]):
    az, pl = azimuth_plunge(eigenvectors[:, i])
    axes.append({
        "Sumbu": name,
        "Azimuth (°)": az,
        "Plunge (°)": pl,
        "Eigenvalue": eigenvalues[i]
    })

axes_df = pd.DataFrame(axes)

print("\nORIENTASI SUMBU TEGASAN")
print(axes_df.round(3).to_string(index=False))

print("\nMISFIT SETIAP EVENT")
print(df[["Event", "Misfit (derajat)"]].round(3).to_string(index=False))

rmse = np.sqrt(np.mean(misfit_deg ** 2))
print(f"\nRMSE misfit = {rmse:.3f} derajat")

# =====================================================
# 6. PLOT DISTRIBUSI STRIKE, DIP, DAN RAKE
# =====================================================
fig, axs = plt.subplots(3, 1, figsize=(9, 10))

parameters = [
    ("Strike", "Strike (°)", (0, 360)),
    ("Dip", "Dip (°)", (0, 90)),
    ("Rake", "Rake (°)", (-180, 180))
]

for ax, (col, label, limits) in zip(axs, parameters):
    ax.bar(df["Event"], df[col], label=label)
    ax.set_ylabel(label)
    ax.set_ylim(*limits)
    ax.set_title(f"Distribusi {col} Delapan Event")
    ax.grid(axis="y", alpha=0.3)
    ax.legend()

axs[-1].set_xlabel("Event gempa")
fig.tight_layout()
fig.savefig(
    "plot_distribusi_mekanisme.png",
    dpi=300, bbox_inches="tight"
)
plt.show()

# =====================================================
# 7. PLOT STEREOGRAFIS HEMISFER BAWAH
# Proyeksi equal-area untuk orientasi sumbu.
# =====================================================
fig, ax = plt.subplots(figsize=(7, 7))

circle = plt.Circle(
    (0, 0), 1, fill=False, color="black"
)
ax.add_patch(circle)

theta = np.linspace(0, 2*np.pi, 361)
ax.plot(np.cos(theta), np.sin(theta), color="black")

symbols = ["$\\sigma_1$", "$\\sigma_2$", "$\\sigma_3$"]

for i, axis in enumerate(axes):
    az = np.radians(axis["Azimuth (°)"])
    pl = np.radians(axis["Plunge (°)"])

    # Equal-area lower-hemisphere projection
    radius = np.sqrt(2) * np.sin(
        (np.pi/2 - pl) / 2
    )

    x = radius * np.sin(az)
    y = radius * np.cos(az)

    ax.scatter(x, y, s=80, label=symbols[i])
    ax.annotate(
        symbols[i], (x, y),
        xytext=(5, 5), textcoords="offset points"
    )

ax.set_aspect("equal")
ax.set_xlim(-1.08, 1.08)
ax.set_ylim(-1.08, 1.08)
ax.set_xlabel("East–West")
ax.set_ylabel("North–South")
ax.set_title("Stereografis Sumbu Tegasan\nHemisfer Bawah")
ax.legend()
ax.grid(alpha=0.25)

fig.savefig(
    "plot_stereografis_tegasan.png",
    dpi=300, bbox_inches="tight"
)
plt.show()

# =====================================================
# 8. PLOT MISFIT PER EVENT
# =====================================================
fig, ax = plt.subplots(figsize=(9, 5))

ax.bar(df["Event"], df["Misfit (derajat)"],
       label="Misfit sudut")

ax.axhline(
    rmse, linestyle="--",
    label=f"RMSE = {rmse:.2f}°"
)

ax.set_xlabel("Event gempa")
ax.set_ylabel("Misfit sudut (°)")
ax.set_title("Misfit Arah Slip untuk Setiap Event")
ax.grid(axis="y", alpha=0.3)
ax.legend()

fig.tight_layout()
fig.savefig(
    "plot_misfit_event.png",
    dpi=300, bbox_inches="tight"
)
plt.show()

print("\nTiga grafik PNG berhasil dibuat jika seluruh proses selesai.")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# SIMULASI SINTETIS - BUKAN DATA PENGUKURAN ASLI
# Zona A: +72 sampai +185 nT
# Zona B: -88 sampai +96 nT
# ==========================================================

# Grid koordinat (km)
x = np.linspace(-8, 8, 321)
y = np.linspace(-8, 8, 321)
X, Y = np.meshgrid(x, y)

# ----------------------------------------------------------
# ZONA A: anomali positif, membulat, dan terlokalisasi
# ----------------------------------------------------------

ZA_awal = np.exp(
    -0.5 * ((X / 2.0)**2 + (Y / 2.0)**2)
)

# Normalisasi agar nilai minimum dan maksimum tepat
ZA = 72 + (ZA_awal - ZA_awal.min()) * (
    (185 - 72) / (ZA_awal.max() - ZA_awal.min())
)

# ----------------------------------------------------------
# ZONA B: anomali bipolar dengan batas memanjang
# Gradien transisi dirancang sekitar 0,7 km
# ----------------------------------------------------------

# Batas magnetik sedikit berkelok, bukan garis sempurna
batas = 0.12 * np.sin(Y / 2.2) + 0.05 * np.sin(Y / 0.9)

# Transisi tajam antara anomali negatif dan positif
transisi = np.tanh((X - batas) / 0.32)

# Variasi kecil sepanjang kelurusan agar tidak seragam
variasi = (
    0.035 * np.sin(Y / 1.8)
    + 0.015 * np.cos(Y / 0.8)
) * np.exp(-((X - batas) / 1.5)**2)

ZB_awal = transisi + variasi

# Normalisasi ke rentang yang diberikan dalam data kasus
ZB = -88 + (ZB_awal - ZB_awal.min()) * (
    (96 - (-88)) / (ZB_awal.max() - ZB_awal.min())
)

# ----------------------------------------------------------
# FUNGSI PEMBUATAN PETA KONTUR
# ----------------------------------------------------------

def plot_kontur(X, Y, Z, judul, nama_file, zona):
    fig, ax = plt.subplots(figsize=(9, 7))

    # Rentang warna mengikuti nilai masing-masing zona
    vmin = float(Z.min())
    vmax = float(Z.max())
    levels = np.linspace(vmin, vmax, 13)

    kontur_warna = ax.contourf(
        X, Y, Z,
        levels=levels,
        cmap="RdBu_r",
        vmin=vmin,
        vmax=vmax
    )

    garis = ax.contour(
        X, Y, Z,
        levels=levels,
        colors="black",
        linewidths=0.65
    )

    # Label hanya pada beberapa garis agar tidak bertumpuk
    ax.clabel(
        garis,
        levels=levels[::2],
        inline=True,
        fontsize=8,
        fmt="%.0f"
    )

    if zona == "A":
        ax.plot(
            0, 0,
            marker="+",
            color="black",
            markersize=12,
            markeredgewidth=2,
            label="Pusat anomali ilustratif"
        )
        ax.legend(loc="upper right", fontsize=8)

    elif zona == "B":
        ax.plot(
            batas[:, 0] if False else np.zeros_like(y),
            y,
            alpha=0
        )

        ax.plot(
            0.12 * np.sin(y / 2.2) + 0.05 * np.sin(y / 0.9),
            y,
            linestyle="--",
            color="black",
            linewidth=1.5,
            label="Batas magnetik ilustratif"
        )
        ax.legend(loc="upper right", fontsize=8)

    ax.set_title(judul, fontsize=12)
    ax.set_xlabel("Jarak X (km)")
    ax.set_ylabel("Jarak Y (km)")
    ax.set_xlim(-8, 8)
    ax.set_ylim(-8, 8)
    ax.set_aspect("equal", adjustable="box")

    cbar = fig.colorbar(kontur_warna, ax=ax, pad=0.03)
    cbar.set_label("Anomali magnetik (nT)")
    cbar.set_ticks(np.linspace(vmin, vmax, 7))

    fig.text(
        0.5, 0.015,
        "SIMULASI SINTETIS — BUKAN DATA PENGUKURAN ASLI",
        ha="center",
        fontsize=9,
        fontweight="bold"
    )

    plt.tight_layout(rect=[0, 0.045, 1, 1])
    plt.savefig(nama_file, dpi=300, bbox_inches="tight")
    plt.show()


# ----------------------------------------------------------
# PLOT 1: ZONA A
# ----------------------------------------------------------

plot_kontur(
    X, Y, ZA,
    "Peta Kontur Anomali Magnetik Sintetis - Zona A",
    "kontur_zona_A.png",
    "A"
)

# ----------------------------------------------------------
# PLOT 2: ZONA B
# ----------------------------------------------------------

plot_kontur(
    X, Y, ZB,
    "Peta Kontur Anomali Magnetik Sintetis - Zona B",
    "kontur_zona_B.png",
    "B"
)

# ----------------------------------------------------------
# PLOT 3: PERBANDINGAN PROFIL PADA Y = 0 KM
# ----------------------------------------------------------

indeks_tengah = np.argmin(np.abs(y - 0))

profil_A = ZA[indeks_tengah, :]
profil_B = ZB[indeks_tengah, :]

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(x, profil_A, label="Zona A", linewidth=2)
ax.plot(x, profil_B, label="Zona B", linewidth=2)

ax.set_title("Perbandingan Profil Anomali Magnetik Sintetis")
ax.set_xlabel("Jarak X (km)")
ax.set_ylabel("Anomali magnetik (nT)")
ax.grid(True, linestyle="--", alpha=0.5)
ax.legend()

fig.text(
    0.5, 0.015,
    "SIMULASI SINTETIS — BUKAN DATA PENGUKURAN ASLI",
    ha="center",
    fontsize=9,
    fontweight="bold"
)

plt.tight_layout(rect=[0, 0.05, 1, 1])
plt.savefig(
    "profil_perbandingan_zona.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()

# ----------------------------------------------------------
# VERIFIKASI RENTANG NILAI
# ----------------------------------------------------------

print(f"Zona A: minimum = {ZA.min():.1f} nT, "
      f"maksimum = {ZA.max():.1f} nT")

print(f"Zona B: minimum = {ZB.min():.1f} nT, "
      f"maksimum = {ZB.max():.1f} nT")

print("Tiga gambar berhasil dibuat jika eksekusi selesai:")
print("1. kontur_zona_A.png")
print("2. kontur_zona_B.png")
print("3. profil_perbandingan_zona.png")
```

Keberadaan sedimen tebal dengan densitas 2,20–2,45 g/cm³ mendukung hipotesis penebalan sedimen, sedangkan basement yang tidak tersingkap membuat kemungkinan pendalaman basement tetap terbuka. Kombinasi penebalan sedimen dan kemungkinan basement yang lebih dalam merupakan model konseptual yang paling menyeluruh berdasarkan informasi yang tersedia, tetapi belum dapat dipastikan sebagai penyebab tunggal tanpa data tambahan mengenai ketebalan sedimen dan geometri basement.

```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# SIMULASI SINTETIS - BUKAN DATA PENGUKURAN ASLI
# ==========================================================

def rescale(data, vmin, vmax):
    return vmin + (data - data.min()) / (
        data.max() - data.min()
    ) * (vmax - vmin)


# Grid koordinat dalam kilometer
x = np.linspace(-20, 20, 301)
y = np.linspace(-15, 15, 241)
X, Y = np.meshgrid(x, y)

# ==========================================================
# PLOT 1: ANOMALI BOUGUER
# Rentang: -42.7 hingga +18.4 mGal
# ==========================================================

bouguer_raw = (
    -1.0 * np.exp(-(Y / 5.0) ** 2)
    - 0.35 * np.exp(
        -(X / 10.0) ** 2 - (Y / 3.2) ** 2
    )
    + 0.18 * np.exp(
        -((X + 12) / 5.0) ** 2
        -((Y - 7) / 4.0) ** 2
    )
    + 0.15 * np.exp(
        -((X - 12) / 5.0) ** 2
        -((Y + 7) / 4.0) ** 2
    )
)

bouguer = rescale(bouguer_raw, -42.7, 18.4)

fig, ax = plt.subplots(figsize=(10, 6.5))
levels = np.linspace(-42.7, 18.4, 17)

cf = ax.contourf(
    X, Y, bouguer,
    levels=levels,
    cmap="RdBu_r"
)

cs = ax.contour(
    X, Y, bouguer,
    levels=levels[::2],
    colors="black",
    linewidths=0.55
)

ax.clabel(cs, fontsize=7, fmt="%.1f")
ax.set_title("Peta Kontur Anomali Bouguer Sintetis")
ax.set_xlabel("Jarak Barat–Timur (km)")
ax.set_ylabel("Jarak Utara–Selatan (km)")
ax.set_aspect("equal", adjustable="box")

cb = fig.colorbar(cf, ax=ax, pad=0.03)
cb.set_label("Anomali Bouguer (mGal)")
cb.set_ticks([-42.7, -30, -15, 0, 18.4])

fig.text(
    0.5, 0.015,
    "SIMULASI SINTETIS — BUKAN DATA PENGUKURAN ASLI",
    ha="center", fontsize=9, fontweight="bold"
)

plt.tight_layout(rect=[0, 0.045, 1, 1])
plt.savefig("anomali_bouguer_sintetis.png",
            dpi=300, bbox_inches="tight")
plt.show()


# ==========================================================
# PLOT 2: ANOMALI RESIDUAL
# Rentang: -17.8 hingga +9.4 mGal
# ==========================================================

residual_raw = (
    -1.0 * np.exp(
        -(X / 5.5) ** 2 - (Y / 3.8) ** 2
    )
    + 0.38 * np.exp(
        -((X - 10) / 4.0) ** 2
        -((Y - 3) / 3.5) ** 2
    )
    + 0.28 * np.exp(
        -((X + 11) / 5.0) ** 2
        -((Y + 5) / 4.0) ** 2
    )
)

residual = rescale(residual_raw, -17.8, 9.4)

fig, ax = plt.subplots(figsize=(9, 6.5))
levels = np.linspace(-17.8, 9.4, 16)

cf = ax.contourf(
    X, Y, residual,
    levels=levels,
    cmap="RdBu_r"
)

cs = ax.contour(
    X, Y, residual,
    levels=levels[::2],
    colors="black",
    linewidths=0.6
)

ax.clabel(cs, fontsize=7, fmt="%.1f")
ax.set_title("Peta Kontur Anomali Residual Sintetis")
ax.set_xlabel("Jarak Barat–Timur (km)")
ax.set_ylabel("Jarak Utara–Selatan (km)")
ax.set_aspect("equal", adjustable="box")

cb = fig.colorbar(cf, ax=ax, pad=0.03)
cb.set_label("Anomali residual (mGal)")
cb.set_ticks([-17.8, -10, 0, 5, 9.4])

fig.text(
    0.5, 0.015,
    "SIMULASI SINTETIS — BUKAN DATA PENGUKURAN ASLI",
    ha="center", fontsize=9, fontweight="bold"
)

plt.tight_layout(rect=[0, 0.045, 1, 1])
plt.savefig("anomali_residual_sintetis.png",
            dpi=300, bbox_inches="tight")
plt.show()


# ==========================================================
# PLOT 3: MODEL KONSEPTUAL PENAMPANG
# Tidak menunjukkan kedalaman sebenarnya
# ==========================================================

fig, axes = plt.subplots(1, 3, figsize=(15, 5.5))

models = [
    ("A. Penebalan Sedimen", 3.0),
    ("B. Basement Lebih Dalam", 3.0),
    ("C. Kombinasi", 4.0)
]

xs = np.array([0, 2, 4, 6, 8, 10])
surface = np.full_like(xs, 5.5, dtype=float)

for ax, (title, center_base) in zip(axes, models):
    # Geometri skematis, bukan kedalaman terukur
    if title.startswith("A"):
        interface = np.array([3.8, 3.8, 2.0, 2.0, 3.8, 3.8])
    elif title.startswith("B"):
        interface = np.array([4.4, 4.0, 2.6, 2.6, 4.0, 4.4])
    else:
        interface = np.array([4.2, 3.8, 1.5, 1.5, 3.8, 4.2])

    ax.fill_between(
        xs, surface, interface,
        color="goldenrod", alpha=0.75,
        label="Sedimen"
    )
    ax.fill_between(
        xs, interface, 0.5,
        color="slategray", alpha=0.75,
        label="Basement kristalin"
    )
    ax.plot(xs, interface, color="black", linewidth=1.6)
    ax.plot([0, 10], [5.5, 5.5], color="black")

    ax.set_title(title, fontsize=10)
    ax.set_xlim(0, 10)
    ax.set_ylim(0.5, 6)
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_xlabel("Jarak horizontal (skematis)")
    ax.set_ylabel("Kedalaman (skematis)")

axes[0].legend(loc="lower left", fontsize=8)

fig.suptitle(
    "Perbandingan Model Konseptual Sumber Anomali Gravitasi",
    fontsize=13
)

fig.text(
    0.5, 0.015,
    "MODEL SKEMATIS — BUKAN HASIL INVERSI; "
    "KEDALAMAN DAN GEOMETRI TIDAK TERUKUR",
    ha="center", fontsize=9, fontweight="bold"
)

plt.tight_layout(rect=[0, 0.06, 1, 0.93])
plt.savefig("model_konseptual_gravitasi.png",
            dpi=300, bbox_inches="tight")
plt.show()

print("Tiga visualisasi selesai dibuat.")
print("1. anomali_bouguer_sintetis.png")
print("2. anomali_residual_sintetis.png")
print("3. model_konseptual_gravitasi.png")

print(f"Rentang Bouguer: {bouguer.min():.1f} hingga "
      f"{bouguer.max():.1f} mGal")
print(f"Rentang residual: {residual.min():.1f} hingga "
      f"{residual.max():.1f} mGal")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# 1. DATA GEMPA
# ============================================================

data = {
    "Event": [
        "E01", "E02", "E03", "E04", "E05",
        "E06", "E07", "E08", "E09", "E10",
        "E11", "E12", "E13", "E14", "E15"
    ],
    "Longitude": [
        119.842, 119.861, 119.879, 119.898, 119.916,
        119.934, 119.953, 119.903, 119.914, 119.926,
        119.938, 119.950, 119.962, 119.974, 119.986
    ],
    "Latitude": [
        -5.218, -5.229, -5.241, -5.253, -5.264,
        -5.275, -5.287, -5.204, -5.220, -5.236,
        -5.251, -5.267, -5.282, -5.298, -5.313
    ],
    "Depth_km": [
        8.2, 9.4, 10.1, 11.3, 12.0,
        12.7, 13.5, 7.1, 7.8, 8.5,
        9.3, 10.0, 10.8, 11.6, 12.4
    ],
    "Magnitude": [
        3.1, 3.4, 3.0, 3.6, 3.2,
        3.5, 3.1, 2.9, 3.2, 3.0,
        3.3, 3.1, 3.4, 3.0, 3.2
    ]
}

df = pd.DataFrame(data)

# ============================================================
# 2. KONVERSI KOORDINAT MENJADI JARAK LOKAL DALAM KM
#    Pendekatan lokal untuk kebutuhan visualisasi
# ============================================================

lon0 = df["Longitude"].mean()
lat0 = df["Latitude"].mean()

df["Easting_km"] = (
    (df["Longitude"] - lon0)
    * 111.32
    * np.cos(np.radians(lat0))
)

df["Northing_km"] = (
    (df["Latitude"] - lat0) * 111.32
)

# Statistik dasar
print("STATISTIK DATA GEMPA")
print(f"Jumlah kejadian : {len(df)}")
print(
    f"Rentang magnitudo: "
    f"{df['Magnitude'].min():.1f}–{df['Magnitude'].max():.1f}"
)
print(
    f"Rentang kedalaman: "
    f"{df['Depth_km'].min():.1f}–{df['Depth_km'].max():.1f} km"
)

# ============================================================
# VISUALISASI 1: PETA SEBARAN EPISENTER
# ============================================================

fig, ax = plt.subplots(figsize=(10, 7))

scatter = ax.scatter(
    df["Longitude"],
    df["Latitude"],
    c=df["Depth_km"],
    s=df["Magnitude"] * 45,
    cmap="viridis",
    edgecolor="black",
    linewidth=0.7
)

for _, row in df.iterrows():
    ax.annotate(
        row["Event"],
        (row["Longitude"], row["Latitude"]),
        xytext=(5, 4),
        textcoords="offset points",
        fontsize=8
    )

cbar = plt.colorbar(scatter, ax=ax)
cbar.set_label("Kedalaman gempa (km)")

ax.set_title("Sebaran Episenter 15 Kejadian Gempa")
ax.set_xlabel("Longitude (°)")
ax.set_ylabel("Latitude (°)")
ax.grid(True, linestyle="--", alpha=0.5)
ax.set_aspect(
    1 / np.cos(np.radians(lat0)),
    adjustable="datalim"
)

plt.tight_layout()
plt.show()

# ============================================================
# VISUALISASI 2: PENAMPANG KEDALAMAN SEPANJANG ARAH UTAMA
# ============================================================

# Analisis komponen utama (PCA) pada koordinat lokal dalam km.
# Ini menentukan sumbu sebaran utama secara geometris,
# bukan menentukan bidang sesar secara langsung.

xy = df[["Easting_km", "Northing_km"]].to_numpy()
xy_centered = xy - xy.mean(axis=0)

eigenvalues, eigenvectors = np.linalg.eigh(
    np.cov(xy_centered.T)
)

main_axis = eigenvectors[:, np.argmax(eigenvalues)]
distance_along_km = xy_centered @ main_axis

df["Distance_along_km"] = distance_along_km

fig, ax = plt.subplots(figsize=(10, 6))

scatter = ax.scatter(
    df["Distance_along_km"],
    df["Depth_km"],
    c=df["Magnitude"],
    s=100,
    cmap="plasma",
    edgecolor="black"
)

for _, row in df.iterrows():
    ax.annotate(
        row["Event"],
        (row["Distance_along_km"], row["Depth_km"]),
        xytext=(4, 4),
        textcoords="offset points",
        fontsize=8
    )

cbar = plt.colorbar(scatter, ax=ax)
cbar.set_label("Magnitudo")

ax.set_title("Penampang Kedalaman Sepanjang Arah Sebaran Utama")
ax.set_xlabel("Jarak proyeksi sepanjang arah utama (km)")
ax.set_ylabel("Kedalaman (km)")
ax.invert_yaxis()
ax.grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()

# ============================================================
# VISUALISASI 3: SEBARAN GEMPA TIGA DIMENSI
# ============================================================

fig = plt.figure(figsize=(11, 8))
ax = fig.add_subplot(111, projection="3d")

scatter = ax.scatter(
    df["Easting_km"],
    df["Northing_km"],
    df["Depth_km"],
    c=df["Magnitude"],
    s=100,
    cmap="plasma",
    edgecolor="black",
    depthshade=True
)

for _, row in df.iterrows():
    ax.text(
        row["Easting_km"],
        row["Northing_km"],
        row["Depth_km"],
        row["Event"],
        fontsize=8
    )

cbar = fig.colorbar(scatter, ax=ax, shrink=0.65, pad=0.1)
cbar.set_label("Magnitudo")

ax.set_title("Visualisasi Tiga Dimensi Sebaran Hiposenter")
ax.set_xlabel("Easting lokal (km)")
ax.set_ylabel("Northing lokal (km)")
ax.set_zlabel("Kedalaman (km)")
ax.invert_zaxis()

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# ==================================================
# DATA
# ==================================================

df = pd.DataFrame({
    "Parameter": [
        "Resistivitas Zona A",
        "Resistivitas Zona B",
        "Gravitasi Zona A",
        "Deformasi horizontal",
        "Jumlah gempa vulkanik"
    ],
    "Januari": [45, 18, -3.2, 1.2, 18],
    "April": [31, 9, -3.8, 7.6, 86],
    "Satuan": ["Ωm", "Ωm", "mGal", "mm/bulan", "event/bulan"]
})

df["Perubahan absolut"] = df["April"] - df["Januari"]

df["Perubahan (%)"] = (
    (df["April"] - df["Januari"]) /
    df["Januari"].abs()
) * 100

# Untuk gravitasi, persentase berdasarkan besar nilai absolut
idx = df["Parameter"] == "Gravitasi Zona A"
df.loc[idx, "Perubahan (%)"] = (
    (abs(-3.8) - abs(-3.2)) / abs(-3.2) * 100
)

print(df.to_string(index=False))

# ==================================================
# PLOT 1: PERBANDINGAN JANUARI DAN APRIL
# ==================================================

fig, axes = plt.subplots(3, 2, figsize=(12, 12))
axes = axes.flatten()

for i, row in df.iterrows():
    ax = axes[i]
    vals = [row["Januari"], row["April"]]

    bars = ax.bar(["Januari", "April"], vals, width=0.55)
    ax.axhline(0, linewidth=0.8)
    ax.set_title(row["Parameter"])
    ax.set_ylabel(row["Satuan"])
    ax.grid(axis="y", linestyle="--", alpha=0.4)

    for bar, val in zip(bars, vals):
        ax.annotate(
            f"{val:g}",
            (bar.get_x() + bar.get_width() / 2, val),
            xytext=(0, 5 if val >= 0 else -5),
            textcoords="offset points",
            ha="center",
            va="bottom" if val >= 0 else "top"
        )

    if row["Parameter"] == "Gravitasi Zona A":
        note = "Perubahan bertanda = -0,6 mGal"
    else:
        note = (
            f"Δ = {row['Perubahan absolut']:+g}; "
            f"{row['Perubahan (%)']:+.2f}%"
        )

    ax.text(
        0.5, 0.92, note,
        transform=ax.transAxes,
        ha="center", va="top", fontsize=9
    )

axes[5].axis("off")

fig.suptitle(
    "Perbandingan Data Geofisika Januari dan April",
    fontsize=15
)
plt.tight_layout()
plt.show()

# ==================================================
# PLOT 2: PENAMPANG KONSEPTUAL
# Kedalaman sesuai data input.
# Posisi horizontal hanya skematis.
# ==================================================

fig, ax = plt.subplots(figsize=(12, 7))
ax.set_xlim(0, 10)
ax.set_ylim(4.5, 0)

# Permukaan
ax.plot([0, 10], [0, 0], linewidth=2)
ax.text(0.2, 0.08, "Permukaan", va="bottom")

# Zona B: 0,3–1 km
zone_b = Rectangle(
    (3.4, 0.3), 3.2, 0.7,
    fill=False, hatch="///", linewidth=1.5
)
ax.add_patch(zone_b)
ax.text(
    5, 0.65,
    "Zona B\n0,3–1 km\nResistivitas turun 50%",
    ha="center", va="center"
)

# Zona A: 1,5–3 km
zone_a = Rectangle(
    (3, 1.5), 4, 1.5,
    fill=False, hatch="\\\\\\", linewidth=1.5
)
ax.add_patch(zone_a)
ax.text(
    5, 2.25,
    "Zona A\n1,5–3 km\nResistivitas turun 31,11%",
    ha="center", va="center"
)

# Deformasi maksimum tepat di atas Zona A
ax.annotate(
    "Deformasi maksimum",
    xy=(5, 0.03),
    xytext=(7.5, 0.55),
    arrowprops=dict(arrowstyle="->"),
    ha="center"
)
ax.text(7.5, 0.8, "7,6 mm/bulan pada April", ha="center", fontsize=9)

# Rentang kedalaman seismicity
ax.axhspan(
    2, 4,
    alpha=0.12,
    hatch="...",
    label="Kedalaman seismicity terbesar: 2–4 km"
)
ax.text(
    1.1, 3.1,
    "Peningkatan\nseismicity terbesar\n2–4 km",
    ha="center", va="center"
)

ax.set_xlabel("Posisi horizontal skematis (bukan skala lokasi)")
ax.set_ylabel("Kedalaman (km)")
ax.set_yticks(np.arange(0, 4.5, 0.5))
ax.set_title("Model Konseptual Perubahan Sistem Bawah Permukaan")
ax.grid(axis="y", linestyle="--", alpha=0.3)
ax.legend(loc="lower right")

ax.text(
    0.02, 0.02,
    "Diagram konseptual, bukan hasil inversi geofisika",
    transform=ax.transAxes,
    fontsize=9
)

plt.tight_layout()
plt.show()

# ==================================================
# PLOT 3: PERSENTASE PERUBAHAN
# ==================================================

labels = [
    "Resistivitas\nZona A",
    "Resistivitas\nZona B",
    "Gravitasi\nZona A*",
    "Deformasi\nhorizontal",
    "Jumlah\ngempa"
]

changes = [
    (31 - 45) / 45 * 100,
    (9 - 18) / 18 * 100,
    (abs(-3.8) - abs(-3.2)) / abs(-3.2) * 100,
    (7.6 - 1.2) / 1.2 * 100,
    (86 - 18) / 18 * 100
]

fig, ax = plt.subplots(figsize=(11, 6))
bars = ax.bar(labels, changes)

ax.axhline(0, linewidth=0.8)
ax.set_ylabel("Perubahan (%)")
ax.set_title("Persentase Perubahan Parameter Januari–April")
ax.grid(axis="y", linestyle="--", alpha=0.4)

for bar, val in zip(bars, changes):
    offset = 7 if val >= 0 else -10
    ax.annotate(
        f"{val:+.2f}%",
        (bar.get_x() + bar.get_width() / 2, val),
        xytext=(0, offset),
        textcoords="offset points",
        ha="center",
        va="bottom" if val >= 0 else "top"
    )

ax.text(
    0.01, 0.02,
    "*Gravitasi dihitung dari besar nilai absolut. "
    "Nilai bertanda berubah dari -3,2 menjadi -3,8 mGal.",
    transform=ax.transAxes,
    fontsize=9
)

plt.tight_layout()
plt.show()
```
